```python
country_code_with_name = {
    'Es':'Spain', 'Ru':'Russia', 'My':'Malaysia', 'Tr':'Turkey',
    'Dk':'Denmark', 'Br':'Brazil', 'Fr':'France', 'Nl':'Netherlands',
    'It':'Italy', 'Cz':'Czech Republic', 'De':'Germany', 'At':'Austria',
    'Gb':'United Kingdom', 'Ge':'Georgia', 'Eg':'Egypt', 'Lt':'Lithuania',
    'Jp':'Japan', 'Be':'Belgium', 'Au':'Australia','Ch':'Switzerland',
    'Ie':'Treland'
}
```

```python
# to check if usable coordinates are there
df_to_fill = df_stat[
    (df_stat['country'].isin(['Es','Ru','My','Tr','Dk','Br','Fr','Nl','It','Cz','De','At','Gb','Ge','Eg','Lt','Jp','Be','Au','Ch','Ie'])) &
    (df_stat[['town','state','postcode']].isna().any(axis=1))
]
```

```python
df_to_fill[['lat','lon']].isna().sum()
```

```python
# Filling one country at a time
es_df = df_stat[
    (df_stat['country'] == 'Es') &
    (df_stat[['town','state','postcode']].isna().any(axis=1))
]
es_df.head()
```

```python
len(es_df)
```

```python
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

geolocator = Nominatim(user_agent='ev_station_cleaning')
reverse = RateLimiter(geolocator.reverse, min_delay_seconds=1)
```

```python
for idx,row in es_df.iterrows():
  location = reverse((row['lat'],row['lon']), language='en')

  if location:
    addr = location.raw.get('address',{})

    if pd.isna(df_stat.at[idx,'town']):
      df_stat.at[idx,'town'] = (
          addr.get('town') or addr.get('city') or addr.get('village'))
    if pd.isna(df_stat.at[idx,'postcode']):
      df_stat.at[idx,'postcode'] = addr.get('postcode')
    if pd.isna(df_stat.at[idx,'state']):
      df_stat.at[idx,'state'] = addr.get('state')
```

```python
for idx,row in df_stat[df_stat['state'].isna()].iterrows():
  location = reverse((row['lat'],row['lon']), language='en')

  if location:
    addr = location.raw.get('address',{})

    if pd.isna(df_stat.at[idx,'state']):
      df_stat.at[idx,'state'] = addr.get('state') or addr.get('region') or addr.get('province') or addr.get('state_district') or addr.get('county')
```
