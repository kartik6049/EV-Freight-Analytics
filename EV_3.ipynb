<h1> Feature Engineering</h1>

```python
cop_df['tons_growth'] = cop_df['tons_2045_pes'] - cop_df['tons_2012']
cop_df['value_growth'] = cop_df['value_2045_pes'] - cop_df['value_2012']
cop_df['tmiles_growth'] = cop_df['tmiles_2045'] - cop_df['tmiles_2012']
```

```python
cop_df.head().T
```

```python
import numpy as np
```

```python
cop_df['tons_growth_perc'] = (
    cop_df['tons_growth'] / cop_df['tons_2012'].replace(0, np.nan)
) * 100
cop_df['value_growth_perc'] = (
    cop_df['value_growth'] / cop_df['value_2012'].replace(0, np.nan)
) * 100
cop_df['tmiles_growth_perc'] = (
    cop_df['tmiles_growth'] / cop_df['tmiles_2012'].replace(0, np.nan)
) * 100
```
